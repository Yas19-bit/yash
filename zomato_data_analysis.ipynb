```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/Zomato data .csv")
df.head()
```

```python
def myrate(value):
  value=str(value).split("/")
  value=value[0]
  return float(value)
df["rate"]=df["rate"].apply(myrate)
df.head()


```

```python
df.isnull().sum()
df.info()
```

type of restorent people order mro


```python
sns.countplot(df["listed_in(type)"])
xlabel="type of resto"
plt.show()
```

```python
sns.lineplot(x="listed_in(type)", y="votes", data=df)
plt.plot()
```

```python
group=df.groupby("listed_in(type)")["votes"].sum()
result=pd.DataFrame({"votes":group})
plt.plot(result,marker="o")
plt.xlabel("resto")
plt.ylabel("votes")
plt.show()
```

highest rate percent


```python
plt.hist(df["rate"], bins=10)
plt.show()
```

couple spendings


```python
couple=df["approx_cost(for two people)"]
sns.countplot(x=couple)
plt.show()
```
